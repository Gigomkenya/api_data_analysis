```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
core_df = pd.read_csv("users_core.csv")
flat_df = pd.read_csv("users_flattened.csv")

print("Core dataset shape:", core_df.shape)
print("Flattened dataset shape:", flat_df.shape)
```

```python
display(core_df.head(10))
```

```python
print("Missing values:")
print(core_df.isnull().sum())

print("\nDuplicate records:", core_df.duplicated().sum())

print("Unique user IDs:", core_df["user_id"].nunique())
```

```python
city_counts = core_df["city"].value_counts()

print(city_counts)

city_counts.plot(kind="bar", figsize=(10, 5))
plt.title("Number of Users by City")
plt.xlabel("City")
plt.ylabel("Number of Users")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
company_counts = core_df["company"].value_counts()

print(company_counts)

company_counts.plot(kind="bar", figsize=(10, 5))
plt.title("Number of Users by Company")
plt.xlabel("Company")
plt.ylabel("Number of Users")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
