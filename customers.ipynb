```python
import pandas as pd
```

# load the extracted data

```python
df = pd.read_csv("extracted_users.csv")
```

# Select customer information

```python
customers = df[[
    "id",
    "name",
    "username",
    "email",
    "address.city",
    "phone",
    "company.name"
]]
print(customers.head())
```

# rename columns

```python
customers = customers.rename(columns={
    "id": "customer_id",
    "name": "customer_name",
    "username": "username",
    "email": "email",
    "address.city": "city",
    "phone": "phone",
    "company.name": "company"
})
print(customers.head())
```

# Export to CSV

```python
customers.to_csv("customers.csv", index=False)
print("customers.csv created successfully!")
print(customers.head())
```
