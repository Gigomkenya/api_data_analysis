```python
import pandas as pd
```

```python
df = pd.read_csv("extracted_users.csv")

print(df.head())
```

3.select important columns 

```python
df = df[[
    "id",
    "name",
    "username",
    "email",
    "address.city",
    "phone",
    "website",
    "company.name"
]]

print(df.head())
```

4.rename the columns

```python
df = df.rename(columns={
    "id": "user_id",
    "name": "full_name",
    "address.city": "city",
    "company.name": "company"
})

print(df.columns)
```

5, check the missing values

```python
print(df.isnull().sum())
```

6.remove duplicates

```python
df = df.drop_duplicates()

print("Duplicates remaining:", df.duplicated().sum())
```

```python
7.Export the cleaned DataFrame
```

```python
df.to_csv("transformed_users.csv", index=False)

print("Transformation completed successfully!")
```
