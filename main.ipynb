# API Data Analysis
## User Data Extraction

```
import json
import requests
import pandas as pd
```

## 1. Retrieve users from the API

```
url = "https://jsonplaceholder.typicode.com/users"
response = requests.get(url)
print(response.status_code)
response.raise_for_status()
```

## 2. Extract JSON data

```
data = response.json()
print(data)
```

## 3. Check the extracted data

```
print(type(data))
print(len(data))
print(data[0])
```

## 4. Convert JSON data into a DataFrame

```
df = pd.json_normalize(data)
print(df.head())
```

## 5. Export extracted data

```
df.to_csv("extracted_users.csv", index=False)
print("Data extraction completed successfully!")
```
