# API Data Analysis Project

## Introduction
This project extracts user information from the JSONPlaceholder API,
transforms nested JSON data into structured datasets, and exports
the results into CSV files.

```python
import json
import requests
import pandas as pd
```

## 1. Retrieve users from the API

```python
url = "https://jsonplaceholder.typicode.com/users"
response = requests.get(url)
print(response.status_code)
response.raise_for_status()
```

## 2. Extract JSON data

```python
data = response.json()
print(data)
```

## 3. Check the extracted data

```python
print(type(data))
print(len(data))
print(data[0])
```

## 4. Convert JSON data into a DataFrame

```python
df = pd.json_normalize(data)
print(df.head())
```

## 5. Export extracted data

```python
data = response.json()

print("Data type:", type(data))
print("Total records:", len(data))

print(data[0])
```

## 6. Trasform dataa into dataframe

```python
df = pd.json_normalize(data)

print("Dataset dimensions:", df.shape)

display(df.head())
```

### 7. Create the core user dataset

```python
core_df = df[
    [
        "id",
        "name",
        "email",
        "address.city",
        "company.name"
    ]
].copy()

core_df.columns = [
    "user_id",
    "name",
    "email",
    "city",
    "company"
]

display(core_df.head())
```

## 8. Export the datasets

```python
df.to_csv("users_flattened.csv", index=False)

core_df.to_csv("users_core.csv", index=False)

print("Both CSV datasets exported successfully!")
```
